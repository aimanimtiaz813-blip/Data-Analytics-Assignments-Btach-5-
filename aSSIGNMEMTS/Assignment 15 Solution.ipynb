# Question 1

## 1. List all cities where the area is greater than 5000, along with their population.

## Solution:

```python
query = """
SELECT 
    c.name AS city_name,
    f.area,
    c.population
FROM cities c
JOIN facts f
    ON c.facts_id = f.id
WHERE f.area > 5000
ORDER BY c.name;
"""

result = pd.read_sql_query(query, conn)

result
```

# Question 2

## List the names of capital cities where the death rate is greater than 2% and the birth rate is less than 20%

## Solution

```python
'''SELECT 
    c.name AS capital_city,
    f.death_rate,
    f.birth_rate
FROM cities c
JOIN facts f
    ON c.facts_id = f.id
WHERE c.capital = 1
  AND f.death_rate > 2
  AND f.birth_rate < 20
ORDER BY c.name;
'''
```

# Question 3

## List the cities with a population growth rate between 1% and 5%, sorted in desc by population growth.

## Solution

```python
"""
SELECT 
    c.name AS city_name,
    f.population_growth
FROM cities c
JOIN facts f
    ON c.facts_id = f.id
WHERE f.population_growth BETWEEN 1 AND 5
ORDER BY f.population_growth DESC;
"""
```

# Question 4

## Find countries that start with "A" and do not have any cities listed in the cities table. The query should return the country name and city name. Using left join

## Solution

```python
"""
SELECT 
    f.name AS country_name,
    c.name AS city_name
FROM facts f
LEFT JOIN cities c
    ON f.id = c.facts_id
WHERE f.name LIKE 'A%'
  AND c.id IS NULL
ORDER BY f.name;
"""
```
