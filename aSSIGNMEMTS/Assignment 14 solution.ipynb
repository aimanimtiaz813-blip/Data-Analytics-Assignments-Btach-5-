```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("database.sqlite")

df = pd.read_sql("SELECT * FROM Salaries", conn)

print("Rows and Columns:", df.shape)

df.head()
```

# Question 1:

 ## What is the average TotalPayBenefits for each JobTitle? exclude ('Not Provided', 'Not provided')

### Solution

```python
query1 = """
SELECT 
    JobTitle,
    AVG(TotalPayBenefits) AS AverageTotalPayBenefits
FROM Salaries
WHERE JobTitle NOT IN ('Not Provided', 'Not provided')
GROUP BY JobTitle
ORDER BY AverageTotalPayBenefits DESC;
"""

q1 = pd.read_sql(query1, conn)
q1
```

# Question 2

## Show The JobTitle and Overtime Pay for All Employees With Overtime Pay greater than 50000

### Solution

```python
query2 = """
SELECT 
    JobTitle,
    OvertimePay
FROM Salaries
WHERE CAST(OvertimePay AS REAL) > 50000
ORDER BY CAST(OvertimePay AS REAL) DESC;
"""

q2 = pd.read_sql(query2, conn)
q2
```

# Question 3

## What is the total number of unique employees? exclude ('Not Provided', 'Not provided')

### Solution

```python
query3 = """
SELECT 
    COUNT(DISTINCT EmployeeName) AS UniqueEmployees
FROM Salaries
WHERE EmployeeName NOT IN ('Not Provided', 'Not provided');
"""

q3 = pd.read_sql(query3, conn)
q3
```

# Question 4

## Show The Top 10 highest Paid Employees

### Solution

```python
query4 = """
SELECT 
    EmployeeName,
    JobTitle,
    TotalPayBenefits
FROM Salaries
ORDER BY TotalPayBenefits DESC
LIMIT 10;
"""

q4 = pd.read_sql(query4, conn)
q4
```

# Question 5

## Show All Employees Ordered By Their TotalPay Benefits In Descending Order

### Solution

```python
query5 = """
SELECT 
    EmployeeName,
    JobTitle,
    TotalPayBenefits
FROM Salaries
ORDER BY TotalPayBenefits DESC;
"""

q5 = pd.read_sql(query5, conn)
q5
```
